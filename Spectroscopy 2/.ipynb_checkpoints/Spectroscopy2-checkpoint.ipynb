# Reducción de Datos Espectroscópicos Parte 2: Calibración en Longitud de Onda

**Resumen:**
En este cuaderno continuamos con la calibración espectral (wavelength calibration). Vamos a:
- cargar imágenes de espectros y convertirlas a vectores de intensidad vs. píxel,
- identificar picos (líneas espectrales) en las imágenes de referencia (He, Ne, Kr, Hg, ...),
- buscar las longitudes de onda conocidas asociadas a esos picos (usando catálogos o la base de datos NIST),
- ajustar un polinomio que convierta píxeles → longitud de onda,
- aplicar esa calibración a espectros observacionales para poder medir características (líneas, anchos, etc.).

**Convertir JPEG a BMP**

Se convierten las imágenes "He.jpeg", "Kr.jpeg", "Hg.jpeg", "Ne.jpeg" al formato bmp

```python
# Cell 2: Convertir JPEG a BMP (opcional)
jpg_files = ["He.jpeg", "Kr.jpeg", "Hg.jpeg", "Ne.jpeg"]
out_format = "bmp"  # o 'png'

for fname in jpg_files:
    if not os.path.exists(fname):
        print(f"Archivo no encontrado: {fname}")
        continue
    base = Path(fname).stem
    outname = f"{base}.{out_format}"
    if not os.path.exists(outname):
        img = Image.open(fname).convert('RGB')
        img.save(outname)
        print(f"Guardado: {outname}")
    else:
        print(f"Ya existe: {outname}")
```

**Se carga la imagen y se convierte a 1D:**

Como las imágenes poseen varios colores se convierten a arreglos unidimensionales para poder procesarlas

```python
# Cell 3: Funciones auxiliares para cargar imagen y extraer perfil espectral
def load_image_gray(path):
    img = Image.open(path).convert('L')
    arr = np.array(img).astype(float)
    arr = (arr - arr.min()) / (arr.max() - arr.min() + 1e-12)
    return arr

def spectral_profile(img_arr, axis=0, crop=None):
    if crop is not None:
        r0,r1,c0,c1 = crop
        img_arr = img_arr[r0:r1, c0:c1]
    return img_arr.sum(axis=axis)

def smooth_profile(profile, window=31, poly=3):
    if len(profile) < window:
        window = max(3, (len(profile)//2)*2+1)
    return savgol_filter(profile, window, poly)
```

```python
hg_gray = load_image_gray(hg_filename)
kr_gray = load_image_gray(kr_filename)
ne_gray = load_image_gray(ne_filename)
he_gray = load_image_gray(he_filename)
```

Nosotros recreamos nuestro modelo de traza del [Tutorial de Trazado Espectroscópico](Spectroscopic%20Trace%20Tutorial.ipynb) usando los modelos ajustados.

Podríamos haber usado la traza empírica directamente, lo cual podría resultar en características de ruido ligeramente mejoradas, pero por simplicidad —y para hacer que ambos cuadernos sean utilizables de manera independiente— usamos aquí los modelos polinómicos y Gaussianos ajustados.


```python
trace_model = Polynomial1D(degree=3, c0=276.03448116, c1=0.0091462, c2=0.00002231, c3=-0.00000001)
trace_profile_model = Gaussian1D(amplitude=123.84846797, mean=0.17719819, stddev=5.10872134)
xaxis = np.arange(he_gray.shape[1])
trace_center = trace_model(xaxis)
npixels_to_cut=15
yaxis = np.arange(-npixels_to_cut, npixels_to_cut)
model_trace_profile = trace_profile_model(yaxis)
```

```python
xaxis_hg = np.arange(hg_gray.shape[1])
trace_center_hg = trace_model(xaxis_hg)
```

```python
xaxis_ne = np.arange(ne_gray.shape[1])
trace_center_ne = trace_model(xaxis_ne)
```

```python
xaxis_kr = np.arange(kr_gray.shape[1])
trace_center_kr = trace_model(xaxis_kr)
```

```python
xaxis_he = np.arange(he_gray.shape[1])
trace_center_he = trace_model(xaxis_he)
```

Luego, usamos la traza para extraer espectros de cada uno de los tres espectros de las lámparas de calibración.

Primero verificamos que las trazas se vean aceptables, ajustamos los límites de los ejes x y y para poder observar el espectro y las trazas de la mejor manera posible

```python
plt.figure(figsize=(16,8))
ax1 = plt.subplot(4,1,1)
ax1.imshow(hg_gray[250:300,:], extent=[0,1000,250,350])
ax1.plot(xaxis_he, trace_center_he)
ax1.set_aspect(3)
ax2 = plt.subplot(4,1,2)
ax2.imshow(kr_gray[250:300,:], extent=[0,1000,250,350])
ax2.plot(xaxis_he, trace_center_he)
ax2.set_aspect(3)
ax3 = plt.subplot(4,1,3)
ax3.imshow(ne_gray[250:300,:], extent=[0,1000,250,350])
ax3.plot(xaxis_he, trace_center_he)
ax3.set_aspect(3)
ax4 = plt.subplot(4,1,4)
ax4.imshow(he_gray[250:300,:], extent=[0,1000,250,350])
ax4.plot(xaxis_he, trace_center_he)
ax4.set_aspect(3)
```

En las siguientes tres celdas, extraemos el espectro promedio ponderado por el perfil de la traza de cada uno de los diferentes elementos que se tienen

``npixels_to_cut`` establece una región recortada alrededor de la traza, similar a lo que se muestra en las figuras anteriores. El ``model_trace_profile``, derivado anteriormente, es el perfil de transmisión que medimos usando la traza estelar.


```python
hg_spectrum = np.array([np.average(hg_gray[int(yval)-npixels_to_cut:int(yval)+npixels_to_cut, ii],
                                 weights=model_trace_profile)
                      for yval, ii in zip(trace_center_hg, xaxis_hg)])
```

```python
ne_spectrum = np.array([np.average(ne_gray[int(yval)-npixels_to_cut:int(yval)+npixels_to_cut, ii],
                                weights=model_trace_profile)
                     for yval, ii in zip(trace_center_ne, xaxis_ne)])
```

```python
kr_spectrum = np.array([np.average(kr_gray[int(yval)-npixels_to_cut:int(yval)+npixels_to_cut, ii],
                                weights=model_trace_profile)
                     for yval, ii in zip(trace_center_kr, xaxis_kr)])
```

```python
he_spectrum = np.array([np.average(he_gray[int(yval)-npixels_to_cut:int(yval)+npixels_to_cut, ii],
                                   weights=model_trace_profile)
                    for yval, ii in zip(trace_center_he, xaxis_he)])
```

Se observan los espectros de los elementos

```python
plt.plot(xaxis_hg, hg_spectrum, label='Mercury')
plt.plot(xaxis_ne, ne_spectrum, label='Neon')
plt.plot(xaxis_kr, kr_spectrum, label='Krypton')
plt.plot(xaxis_he, he_spectrum, label='Helium')
plt.legend(loc='best');
```

En este siguiente paso, tenemos que anotar qué valores (aproximados) de píxel corresponden a longitudes de onda conocidas.

Adivinar esto simplemente mirando los espectros y comparándolos con atlas espectrales es difícil y requiere experiencia.

El siguiente truco es averiguar en qué dirección aumenta la longitud de onda: ¿hacia la izquierda o hacia la derecha? El mercurio no ayuda con eso, ya que los dobletes azul-violeta y verde-amarillo están aproximadamente a la misma distancia entre sí. Resulta que, a partir de trabajo de prueba y error, la longitud de onda aumenta hacia la derecha.

En la siguiente línea de código lo que se hace es ubicar los pixeles que se van a estudiar y las longitudes de onda teóricas que se tienen para dichas líneas de emisión, tener presente que deben ser la misma cantidad (las longitudes de onda se encuentran en nm)

***Mercurio***

```python
guessed_wavelengths_hg = [435.49, 491.23, 545.82, 567.65, 587.19, 606.73]
guessed_xvals_hg = [100, 203, 302, 363, 408, 472]
```

## Mejorando nuestras suposiciones

Podemos hacerlo mucho mejor al determinar los valores X de píxel tomando la coordenada ponderada por la intensidad (momento 1):

```python
npixels = 15
improved_xval_guesses_hg = [np.average(xaxis_hg[g-npixels:g+npixels],
                                    weights=hg_spectrum[g-npixels:g+npixels] - np.median(hg_spectrum))
                         for g in guessed_xvals_hg]
improved_xval_guesses_hg
```

Observamos los valores de nuestras supocisiones y los mejorados, si coiciden con los picos seleccionados se está realizando de forma óptima, si quedan muy lejos hay que cambiar el valor del pixel de nuestras suposiciones

```python
plt.plot(xaxis_hg, hg_spectrum, label='Espectro Hg')
plt.plot(guessed_xvals_hg, [0.5]*6, 'rx', label='Estimación inicial')
plt.plot(improved_xval_guesses_hg, [0.5]*6, 'g+', label='Centro refinado')
plt.legend()
plt.xlabel('Píxel')
plt.ylabel('Intensidad')
plt.show()
```

Podemos consultar las listas de líneas de neón, kriptón y helio para ver qué tan bien lo hicimos y quizá intentar mejorar nuestro ajuste.

NIST, el Instituto Nacional de Estándares y Tecnología, mantiene [listas de líneas atómicas](https://physics.nist.gov/PhysRefData/ASD/lines_form.html). Sin embargo, ¡no tenemos una manera directa de saber qué transiciones de estos átomos usar! Aun así, podemos aplicar algunas heurísticas.

```python
# we adopt the minimum/maximum wavelength from our linear fit
minwave_hg = wavelengths_hg.min()
maxwave_hg = wavelengths_hg.max()

minwave_ne = wavelengths_ne.min()
maxwave_ne = wavelengths_ne.max()

minwave_kr = wavelengths_kr.min()
maxwave_kr = wavelengths_kr.max()

minwave_he = wavelengths_he.min()
maxwave_he = wavelengths_he.max()

# then we search for atomic lines
# We are only interested in neutral lines, assuming the lamps are not hot enough to ionize the atoms
mercury_lines = Nist.query(minwav=minwave_hg,
                           maxwav=maxwave_hg,
                           linename='Hg I')
krypton_lines = Nist.query(minwav=minwave_kr,
                           maxwav=maxwave_kr,
                           linename='Kr I')
neon_lines = Nist.query(minwav=minwave_ne,
                        maxwav=maxwave_ne,
                        linename='Ne I')
helium_lines = Nist.query(minwav=minwave_he,
                         maxwav=maxwave_he,
                         linename='He I')
```

Podemos comenzar mirando lo que NIST reporta para el espectro de mercurio que ya ajustamos:
